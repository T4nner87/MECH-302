```python
import pandas as pd
import matplotlib.pyplot as plt
import re

# Enter the path to your PuTTY log file
file_path = r"C:\Users\YourName\Downloads\encoder.log"

# Read the log file
with open(file_path, "r", errors="ignore") as file:
    lines = file.readlines()

print(f"Read {len(lines)} lines from the log file.")

# Pattern matching the Arduino serial output
pattern = (
    r"Count\s*=\s*(-?\d+)"
    r".*?Angle\s*=\s*(-?[\d.]+)\s*rad"
    r".*?Angular Velocity\s*=\s*(-?[\d.]+)\s*rad/s"
    r".*?Angular Acceleration\s*=\s*(-?[\d.]+)\s*rad/s\^2"
)

data = []

for line in lines:
    match = re.search(pattern, line)

    if match:
        count, angle, velocity, acceleration = match.groups()

        data.append([
            int(count),
            float(angle),
            float(velocity),
            float(acceleration)
        ])

# Create a DataFrame
df = pd.DataFrame(data, columns=[
    "Encoder Count",
    "Angle (rad)",
    "Angular Velocity (rad/s)",
    "Angular Acceleration (rad/s^2)"
])

print(f"Extracted {len(df)} valid data points.")
df.head()
```
