```python
import time
import itertools
from pathlib import Path

import serial
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# =====================================================
# USER SETTINGS
# =====================================================

PORT = "COM3"       # Change to your Arduino COM port
BAUD = 115200

EXCITATION_MM = 20.0
FEEDBACK_DURATION_S = 10.0

SERIAL_TIMEOUT = 3.0

OUTPUT_DIR = Path("closed_loop_results")
OUTPUT_DIR.mkdir(exist_ok=True)


# =====================================================
# SELECT EXPERIMENT
# =====================================================
TASK = input("please enter the task you would like to run from the list: maintain, quash, frequency")

# Theoretical pendulum frequency
OMEGA0 = np.sqrt(9.81 / 0.120)

# Desired closed-loop damping ratio for quashing
ZETA_DESIRED = 0.5

# Gain grids
TASK_GRIDS = {

    "quash": {
        "kp": [OMEGA0**2],
        "kd": [
            2 * ZETA_DESIRED * OMEGA0,
            2 * 0.7 * OMEGA0,
            2 * 1.0 * OMEGA0
        ]
    },

    "maintain": {
        "kp": [OMEGA0**2],
        "kd": [0.0, 0.25, 0.5]
    },

    "frequency": {
        "kp": [
            1.25 * OMEGA0**2,
            1.50 * OMEGA0**2,
            2.00 * OMEGA0**2
        ],
        "kd": [0.0, 0.25, 0.5]
    }
}


# =====================================================
# SERIAL FUNCTIONS
# =====================================================

def read_line(ser, timeout=30):

    deadline = time.time() + timeout

    while time.time() < deadline:

        raw = ser.readline()

        if not raw:
            continue

        line = raw.decode(
            errors="replace"
        ).strip()

        if line:
            print(line)

            if line.startswith("ERR,"):
                raise RuntimeError(line)

            return line

    raise TimeoutError("Arduino response timed out")


def wait_for_prefix(ser, prefix, timeout=10):

    deadline = time.time() + timeout

    while time.time() < deadline:

        line = read_line(
            ser,
            max(0.1, deadline - time.time())
        )

        if line.startswith(prefix):
            return line

    raise TimeoutError(
        f"Did not receive {prefix}"
    )


# =====================================================
# RUN ONE TRIAL
# =====================================================

def run_trial(ser, kp, kd, trial_number):

    command = (
        f"TRIAL,{EXCITATION_MM:.3f},"
        f"{kp:.6f},{kd:.6f},"
        f"{FEEDBACK_DURATION_S:.3f}\n"
    )

    ser.write(command.encode())
    ser.flush()

    wait_for_prefix(ser, "BEGIN", timeout=15)

    records = []

    while True:

        line = read_line(ser, timeout=30)

        if line == "END":
            break

        if line.startswith("DATA,"):

            fields = line.split(",")

            if len(fields) == 6:

                records.append({

                    "time_s": float(fields[1]),
                    "cart_mm": float(fields[2]),
                    "theta_deg": float(fields[3]),
                    "theta_dot_deg_s": float(fields[4]),
                    "commanded_velocity_mm_s": float(fields[5])

                })

    wait_for_prefix(
        ser,
        "RETURNED",
        timeout=30
    )

    df = pd.DataFrame(records)

    if df.empty:
        raise RuntimeError("No trial data received")

    # Add trial information
    df["trial"] = trial_number
    df["Kp"] = kp
    df["Kd"] = kd
    df["task"] = TASK

    # Save data
    filename = (
        OUTPUT_DIR /
        f"{TASK}_trial_{trial_number:02d}.csv"
    )

    df.to_csv(filename, index=False)

    # -------------------------------------------------
    # Calculate response metrics
    # -------------------------------------------------

    theta = df["theta_deg"].to_numpy()
    time_s = df["time_s"].to_numpy()

    rms_angle = np.sqrt(np.mean(theta**2))
    peak_angle = np.max(np.abs(theta))

    # Estimate frequency using positive-going zero crossings
    mask = time_s >= 1.0

    t = time_s[mask]
    y = theta[mask]

    crossings = []

    for i in range(1, len(y)):

        if y[i-1] <= 0 and y[i] > 0:

            fraction = (
                -y[i-1] / (y[i] - y[i-1])
            )

            crossing_time = (
                t[i-1] +
                fraction * (t[i] - t[i-1])
            )

            crossings.append(crossing_time)

    if len(crossings) >= 2:

        frequency = (
            1.0 / np.mean(np.diff(crossings))
        )

    else:

        frequency = np.nan

    result = {

        "trial": trial_number,
        "Kp": kp,
        "Kd": kd,
        "RMS_angle_deg": rms_angle,
        "Peak_angle_deg": peak_angle,
        "Frequency_Hz": frequency,
        "samples": len(df),
        "filename": str(filename)

    }

    return df, result


# =====================================================
# MAIN EXPERIMENT
# =====================================================

def main():

    if TASK not in TASK_GRIDS:
        raise ValueError(
            "TASK must be quash, maintain, or frequency"
        )

    kp_values = TASK_GRIDS[TASK]["kp"]
    kd_values = TASK_GRIDS[TASK]["kd"]

    combinations = list(
        itertools.product(kp_values, kd_values)
    )

    summary = []

    print("\nCART-PENDULUM CLOSED-LOOP TEST")
    print("--------------------------------")

    print("Task:", TASK)
    print("Natural frequency:", OMEGA0, "rad/s")
    print("Number of trials:", len(combinations))

    with serial.Serial(
        PORT,
        BAUD,
        timeout=SERIAL_TIMEOUT
    ) as ser:

        time.sleep(2)

        ser.reset_input_buffer()

        ser.write(b"PING\n")

        wait_for_prefix(
            ser,
            "READY",
            timeout=5
        )

        for i, (kp, kd) in enumerate(
            combinations,
            start=1
        ):

            print("\n--------------------------------")
            print(f"TRIAL {i}/{len(combinations)}")
            print("Kp =", kp)
            print("Kd =", kd)

            input(
                "\nCentre the cart, let the pendulum "
                "hang still, and press Enter..."
            )

            ser.write(b"ZERO\n")

            wait_for_prefix(
                ser,
                "OK,ZERO",
                timeout=5
            )

            df, result = run_trial(
                ser,
                kp,
                kd,
                i
            )

            summary.append(result)

            pd.DataFrame(summary).to_csv(
                OUTPUT_DIR / f"{TASK}_summary.csv",
                index=False
            )

            # Plot individual trial
            plt.figure(figsize=(9, 5))

            plt.plot(
                df["time_s"],
                df["theta_deg"],
                label=f"Kp={kp:.2f}, Kd={kd:.2f}"
            )

            plt.axhline(0, linewidth=0.8)

            plt.xlabel("Time [s]")
            plt.ylabel("Pendulum angle [degrees]")

            plt.title(
                f"{TASK.capitalize()} - Trial {i}"
            )

            plt.grid(True)
            plt.legend()
            plt.tight_layout()

            plt.savefig(
                OUTPUT_DIR / f"{TASK}_trial_{i:02d}.png"
            )

            plt.show()

    # -------------------------------------------------
    # Summary
    # -------------------------------------------------

    results = pd.DataFrame(summary)

    print("\nFINAL RESULTS")
    print(results.to_string(index=False))

    results.to_csv(
        OUTPUT_DIR / f"{TASK}_summary.csv",
        index=False
    )

    # Gain comparison
    plt.figure(figsize=(8, 5))

    scatter = plt.scatter(
        results["Kp"],
        results["Kd"],
        c=results["RMS_angle_deg"],
        s=120
    )

    plt.colorbar(
        scatter,
        label="RMS angle [degrees]"
    )

    plt.xlabel("Kp [s^-2]")
    plt.ylabel("Kd [s^-1]")

    plt.title(
        f"{TASK.capitalize()} - Gain comparison"
    )

    plt.grid(True)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / f"{TASK}_gain_comparison.png"
    )

    plt.show()


if __name__ == "__main__":
    main()
```

```python

"""
CART-PENDULUM FEEDBACK EXPERIMENT RUNNER

Arduino commands:
    PING
    ZERO
    TRIAL,MODE,excitation_mm,Kp,Kd,duration

Tasks:
    quash
    maintain
    frequency

Runs:
    1. No-feedback hardware trial
    2. Selected feedback hardware trial
    3. Simulated response
    4. Comparison plot
"""

import time
from pathlib import Path

import serial
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. USER SETTINGS
# ============================================================

PORT = "COM3"        # Change to your Arduino COM port
BAUD = 115200

TASK = input(
    "Select task (quash, maintain, frequency): "
).strip().lower()

EXCITATION_MM = 30.0
FEEDBACK_DURATION_S = 10.0

OUTPUT_DIR = Path("cart_pendulum_results")
OUTPUT_DIR.mkdir(exist_ok=True)


# ============================================================
# 2. MODEL PARAMETERS
# ============================================================

G = 9.81

NATURAL_FREQ_HZ = 1.36
OMEGA_N = 2.0 * np.pi * NATURAL_FREQ_HZ

ZETA = 0.05

NATURAL_DAMPING = 2.0 * ZETA * OMEGA_N

MAX_ACCEL_M_S2 = 2.621
MAX_SPEED_M_S = 0.1634
MAX_CART_DISPLACEMENT_M = 0.070

ENCODER_COUNTS_PER_REV = 2400


# ============================================================
# 3. FEEDBACK GAINS
# ============================================================

NATURAL_KP = OMEGA_N**2

GAIN_SETTINGS = {

    "quash": {
        "Kp": -5.284,
        "Kd": 0.953
    },

    "maintain": {
        "Kp": 0,
        "Kd": -0.0101
    },

    "frequency": {
        "Kp": 6.702,
        "Kd": -0.0101
    }
}

if TASK not in GAIN_SETTINGS:
    raise ValueError(
        "TASK must be quash, maintain, or frequency."
    )

KP = GAIN_SETTINGS[TASK]["Kp"]
KD = GAIN_SETTINGS[TASK]["Kd"]


# ============================================================
# 4. SERIAL COMMUNICATION
# ============================================================

def read_line(ser, timeout=30):

    deadline = time.time() + timeout

    while time.time() < deadline:

        raw = ser.readline()

        if not raw:
            continue

        line = raw.decode(
            errors="replace"
        ).strip()

        if line:

            print(line)

            if line.startswith(("ERR", "ERROR")):
                raise RuntimeError(
                    f"Arduino reported: {line}"
                )

            return line

    raise TimeoutError(
        "Timed out waiting for Arduino."
    )


def wait_for_message(ser, message, timeout=10):

    deadline = time.time() + timeout

    while time.time() < deadline:

        remaining = deadline - time.time()

        line = read_line(
            ser,
            timeout=max(0.1, remaining)
        )

        if line == message:
            return

    raise TimeoutError(
        f"Did not receive {message}."
    )


# ============================================================
# 5. HARDWARE TRIAL
# ============================================================

def run_hardware_trial(ser, mode, kp, kd):

    command = (
        f"TRIAL,{mode.upper()},"
        f"{EXCITATION_MM:.3f},"
        f"{kp:.6f},"
        f"{kd:.6f},"
        f"{FEEDBACK_DURATION_S:.3f}\n"
    )

    print("\nSending command:")
    print(command.strip())

    ser.write(command.encode("utf-8"))
    ser.flush()

    wait_for_message(
        ser,
        "BEGIN",
        timeout=10
    )

    records = []

    while True:

        line = read_line(
            ser,
            timeout=30
        )

        if line == "END":
            break

        if line.startswith("DATA,"):

            fields = line.split(",")

            if len(fields) != 7:
                continue

            records.append({

                "time_s": float(fields[1]),

                "stage": fields[2],

                "cart_mm": float(fields[3]),

                "theta_rad": float(fields[4]),

                "theta_dot_rad_s": float(fields[5]),

                "target_mm": float(fields[6])
            })

    wait_for_message(
        ser,
        "RETURNED",
        timeout=60
    )

    df = pd.DataFrame(records)

    if df.empty:
        raise RuntimeError(
            "No data received from Arduino."
        )

    return df


# ============================================================
# 6. GET INITIAL CONDITIONS
# ============================================================

def get_feedback_initial_condition(df):

    control_indices = np.where(
        df["stage"].to_numpy() == "CONTROL"
    )[0]

    if len(control_indices) == 0:
        raise RuntimeError(
            "No CONTROL data found."
        )

    first_control = control_indices[0]

    theta0 = df.iloc[first_control]["theta_rad"]

    omega0 = df.iloc[first_control]["theta_dot_rad_s"]

    t0 = df.iloc[first_control]["time_s"]

    return first_control, t0, theta0, omega0


# ============================================================
# 7. SIMULATION
# ============================================================

def simulate_response(df, kp, kd):

    first_control, control_start, theta0, omega0 = (
        get_feedback_initial_condition(df)
    )

    time_data = df["time_s"].to_numpy()

    measured_angle = df["theta_rad"].to_numpy()

    theta_sim = np.full(
        len(time_data),
        np.nan
    )

    # Use measured data before feedback begins.
    theta_sim[:first_control] = (
        measured_angle[:first_control]
    )

    theta = float(theta0)
    omega = float(omega0)

    x = 0.0
    v = 0.0

    last_time = control_start

    K = OMEGA_N**2 / G

    for i in range(first_control, len(time_data)):

        current_time = time_data[i]

        dt = current_time - last_time

        if dt <= 0:

            theta_sim[i] = theta

            continue

        # Cart acceleration gains, matching the Arduino sketch.
        cart_accel = kp * theta + kd * omega

        cart_accel = np.clip(
            cart_accel,
            -MAX_ACCEL_M_S2,
            MAX_ACCEL_M_S2
        )

        # Approximate cart motion.
        v += cart_accel * dt

        v = np.clip(
            v,
            -MAX_SPEED_M_S,
            MAX_SPEED_M_S
        )

        x += v * dt

        if x >= MAX_CART_DISPLACEMENT_M:

            x = MAX_CART_DISPLACEMENT_M
            v = 0.0
            cart_accel = 0.0

        elif x <= -MAX_CART_DISPLACEMENT_M:

            x = -MAX_CART_DISPLACEMENT_M
            v = 0.0
            cart_accel = 0.0

        # Pendulum equation.
        theta_ddot = (

            -(OMEGA_N**2) * theta

            - NATURAL_DAMPING * omega

            - (OMEGA_N**2 / G) * cart_accel
        )

        # Euler integration.
        omega += theta_ddot * dt

        theta += omega * dt

        theta_sim[i] = theta

        last_time = current_time

    return theta_sim, control_start


# ============================================================
# 8. FREQUENCY ESTIMATION
# ============================================================

def estimate_frequency(time_data, angle_data):

    time_data = np.asarray(time_data)
    angle_data = np.asarray(angle_data)

    crossings = []

    for i in range(1, len(angle_data)):

        if angle_data[i - 1] <= 0 < angle_data[i]:

            difference = (
                angle_data[i] - angle_data[i - 1]
            )

            if difference == 0:
                continue

            fraction = (
                -angle_data[i - 1] / difference
            )

            crossing_time = (
                time_data[i - 1]
                +
                fraction * (
                    time_data[i] - time_data[i - 1]
                )
            )

            crossings.append(crossing_time)

    if len(crossings) < 3:
        return np.nan

    periods = np.diff(crossings)

    return 1.0 / np.mean(periods)


# ============================================================
# 9. INITIAL AMPLITUDE
# ============================================================

def get_initial_amplitude(df):

    free_data = df[
        df["stage"] == "FREE"
    ]

    if len(free_data) == 0:

        raise RuntimeError(
            "No free-swing measurements available."
        )

    amplitude = np.max(
        np.abs(
            free_data["theta_rad"].to_numpy()
        )
    )

    return amplitude


# ============================================================
# 10. COMPARISON PLOT
# ============================================================

def create_comparison_plot(
    control_df,
    feedback_df
):

    theta_sim, control_start = simulate_response(
        feedback_df,
        KP,
        KD
    )

    initial_amplitude = get_initial_amplitude(
        feedback_df
    )

    fig, ax = plt.subplots(figsize=(12, 6))

    # --------------------------------------------------------
    # NO-FEEDBACK HARDWARE
    # --------------------------------------------------------

    ax.plot(
        control_df["time_s"],
        np.rad2deg(control_df["theta_rad"]),
        label="No-feedback hardware",
        linewidth=1.5
    )

    # --------------------------------------------------------
    # SIMULATION
    # --------------------------------------------------------

    ax.plot(
        feedback_df["time_s"],
        np.rad2deg(theta_sim),
        "--",
        label="Closed-loop simulation",
        linewidth=2
    )

    # --------------------------------------------------------
    # FEEDBACK HARDWARE
    # --------------------------------------------------------

    ax.plot(
        feedback_df["time_s"],
        np.rad2deg(feedback_df["theta_rad"]),
        label="Closed-loop hardware",
        linewidth=1.5
    )

    # --------------------------------------------------------
    # FEEDBACK START
    # --------------------------------------------------------

    ax.axvline(
        control_start,
        linestyle=":",
        linewidth=1.5,
        label="Feedback begins"
    )

    # --------------------------------------------------------
    # TASK-SPECIFIC ANNOTATIONS
    # --------------------------------------------------------

    if TASK == "quash":

        target = 0.10 * initial_amplitude

        ax.axhline(
            np.rad2deg(target),
            color="red",
            linestyle=":",
            label="10% amplitude target"
        )

        ax.axhline(
            -np.rad2deg(target),
            color="red",
            linestyle=":"
        )

        ax.set_title(
            "Task 1: Quash Pendulum Oscillations"
        )

    elif TASK == "maintain":

        ax.axhline(
            np.rad2deg(initial_amplitude),
            linestyle=":",
            label="Initial amplitude"
        )

        ax.axhline(
            -np.rad2deg(initial_amplitude),
            linestyle=":"
        )

        ax.set_title(
            "Task 2: Maintain Pendulum Oscillations"
        )

    elif TASK == "frequency":

        ax.axhline(
            np.rad2deg(initial_amplitude),
            linestyle=":",
            label="Initial amplitude"
        )

        ax.axhline(
            -np.rad2deg(initial_amplitude),
            linestyle=":"
        )

        ax.set_title(
            "Task 3: Tune Pendulum Frequency"
        )

    ax.set_xlabel(
        "Time from start of cart movement [s]"
    )

    ax.set_ylabel(
        "Pendulum angle [degrees]"
    )

    ax.grid(True, alpha=0.3)

    ax.legend()

    fig.tight_layout()

    filename = (
        OUTPUT_DIR / f"{TASK}_comparison.png"
    )

    fig.savefig(
        filename,
        dpi=200
    )

    plt.show()

    # --------------------------------------------------------
    # FREQUENCY MEASUREMENTS
    # --------------------------------------------------------

    feedback_data = feedback_df[
        feedback_df["stage"] == "CONTROL"
    ]

    hardware_frequency = estimate_frequency(
        feedback_data["time_s"],
        feedback_data["theta_rad"]
    )

    simulation_indices = np.where(
        feedback_df["time_s"].to_numpy() >= control_start
    )[0]

    simulation_frequency = estimate_frequency(
        feedback_df["time_s"].to_numpy()[simulation_indices],
        theta_sim[simulation_indices]
    )

    # --------------------------------------------------------
    # RESULTS
    # --------------------------------------------------------

    print("\n---------------- RESULTS ----------------")

    print(f"Task: {TASK}")

    print(f"Kp: {KP:.4f} s^-2")

    print(f"Kd: {KD:.4f} s^-1")

    print(
        "Initial amplitude: "
        f"{np.rad2deg(initial_amplitude):.3f} degrees"
    )

    print(
        f"Hardware frequency: {hardware_frequency:.4f} Hz"
    )

    print(
        f"Simulation frequency: {simulation_frequency:.4f} Hz"
    )

    print(
        f"Natural frequency: {NATURAL_FREQ_HZ:.4f} Hz"
    )

    print(f"Figure saved to: {filename}")

    return theta_sim


# ============================================================
# 11. MAIN
# ============================================================

def main():

    print("\nCART-PENDULUM EXPERIMENT")
    print("-----------------------")

    print(f"Task: {TASK}")

    print(f"Kp = {KP:.4f} s^-2")

    print(f"Kd = {KD:.4f} s^-1")

    print(
        f"Natural frequency = {NATURAL_FREQ_HZ:.3f} Hz"
    )

    print(
        f"Excitation distance = {EXCITATION_MM:.1f} mm"
    )

    with serial.Serial(
        PORT,
        BAUD,
        timeout=1
    ) as ser:

        # Allow Arduino to reset after serial connection.
        time.sleep(3)

        ser.reset_input_buffer()

        # ----------------------------------------------------
        # CHECK CONNECTION
        # ----------------------------------------------------

        print("\nConnecting to Arduino...")

        ser.write(b"PING\n")
        ser.flush()

        wait_for_message(
            ser,
            "READY",
            timeout=10
        )

        print("Arduino connected successfully.")

        # ----------------------------------------------------
        # CONTROL CASE
        # ----------------------------------------------------

        print("\nNO-FEEDBACK CONTROL CASE")

        input(
            "Centre the cart and hang the pendulum vertically. "
            "Press Enter when ready..."
        )

        ser.write(b"ZERO\n")
        ser.flush()

        wait_for_message(
            ser,
            "OK,ZERO",
            timeout=5
        )

        control_df = run_hardware_trial(
            ser,
            "CONTROL",
            0.0,
            0.0
        )

        control_df.to_csv(
            OUTPUT_DIR / "control_case.csv",
            index=False
        )

        print("Control data saved.")

        # ----------------------------------------------------
        # FEEDBACK CASE
        # ----------------------------------------------------

        print("\nFEEDBACK CASE")

        print(f"Running {TASK.upper()} task.")

        input(
            "Centre the cart again and hang the pendulum "
            "vertically. Press Enter when ready..."
        )

        ser.write(b"ZERO\n")
        ser.flush()

        wait_for_message(
            ser,
            "OK,ZERO",
            timeout=5
        )

        feedback_df = run_hardware_trial(
            ser,
            TASK,
            KP,
            KD
        )

        feedback_df.to_csv(
            OUTPUT_DIR / f"{TASK}_hardware.csv",
            index=False
        )

        print("Feedback data saved.")

        # ----------------------------------------------------
        # SAVE GAINS
        # ----------------------------------------------------

        pd.DataFrame([{

            "task": TASK,

            "Kp_s-2": KP,

            "Kd_s-1": KD,

            "natural_frequency_Hz": NATURAL_FREQ_HZ,

            "damping_ratio_assumed": ZETA,

            "excitation_mm": EXCITATION_MM

        }]).to_csv(
            OUTPUT_DIR / f"{TASK}_gains.csv",
            index=False
        )

        # ----------------------------------------------------
        # CREATE PLOT
        # ----------------------------------------------------

        create_comparison_plot(
            control_df,
            feedback_df
        )

    print("\nExperiment complete.")


if __name__ == "__main__":
    main()
```
